```python
import numpy as np
import matplotlib.pyplot as plt

print("CDFs and PDFs")

# 1. Piecewise sqrt(x)
def F1(x):
    return np.where(x <= 0, 0,
           np.where(x >= 1, 1, np.sqrt(x)))

def f1(x):
    """Density: f(x) = 1/(2*sqrt(x)) for 0 < x < 1"""
    return np.where((x > 0) & (x < 1), 1/(2*np.sqrt(x)), 0)

# 2. Logistic CDF
def F2(x):
    return 1 / (1 + np.exp(-x))

def f2(x):
    """Density: f(x) = e^(-x)/(1+e^(-x))^2"""
    exp_neg_x = np.exp(-x)
    return exp_neg_x / (1 + exp_neg_x)**2

# 3. Custom quadratic (for a < c < b)
a, c, b = 0, 1, 2  # Example parameters
def F3(x):
    res = np.zeros_like(x)
    # between a and c
    mask1 = (x >= a) & (x <= c)
    res[mask1] = ((x[mask1] - a)**2) / ((b - a) * (c - a))
    # between c and b
    mask2 = (x > c) & (x < b)
    res[mask2] = 1 - ((b - x[mask2])**2) / ((b - a) * (b - c))
    # above b
    res[x >= b] = 1
    return res

def f3(x):
    """Density: piecewise linear (triangular-like)"""
    res = np.zeros_like(x)
    mask1 = (x > a) & (x < c)
    res[mask1] = 2 * (x[mask1] - a) / ((b - a) * (c - a))
    mask2 = (x > c) & (x < b)
    res[mask2] = 2 * (b - x[mask2]) / ((b - a) * (b - c))
    return res

# 4. Exponential CDF
def F4(x):
    return np.where(x < 0, 0, 1 - np.exp(-x))

def f4(x):
    """Density: f(x) = e^(-x) for x > 0"""
    return np.where(x > 0, np.exp(-x), 0)

# 5. Uniform(0,1) CDF
def F5(x):
    return np.where(x <= 0, 0,
           np.where(x >= 1, 1, x))

def f5(x):
    """Density: f(x) = 1 for 0 < x < 1"""
    return np.where((x > 0) & (x < 1), 1, 0)

# Plotting
x = np.linspace(-2, 3, 400)
fig, axs = plt.subplots(5, 2, figsize=(12, 12))

funcs_cdf = [F1, F2, F3, F4, F5]
funcs_pdf = [f1, f2, f3, f4, f5]
titles = [
    r"1. $F(x) = \sqrt{x}$ for $0 \leq x \leq 1$",
    r"2. $F(x) = \frac{1}{1+e^{-x}}$ (Logistic)",
    r"3. Triangular-like (a=0, c=1, b=2)",
    r"4. $F(x) = 1 - e^{-x}$ (Exponential)",
    r"5. $F(x) = x$ (Uniform)"
]

for i, (f_cdf, f_pdf, title) in enumerate(zip(funcs_cdf, funcs_pdf, titles)):
    # Plot CDF
    axs[i, 0].plot(x, f_cdf(x), linewidth=2, color='blue')
    axs[i, 0].set_ylim(-0.1, 1.1)
    axs[i, 0].set_xlim(-2, 3)
    axs[i, 0].grid(True, alpha=0.3)
    axs[i, 0].set_title(f"{title} - CDF", fontsize=10)
    axs[i, 0].set_xlabel("x")
    axs[i, 0].set_ylabel("F(x)")
    
    # Plot PDF
    axs[i, 1].plot(x, f_pdf(x), linewidth=2, color='red')
    axs[i, 1].set_xlim(-2, 3)
    axs[i, 1].grid(True, alpha=0.3)
    axs[i, 1].set_title(f"{title} - PDF", fontsize=10)
    axs[i, 1].set_xlabel("x")
    axs[i, 1].set_ylabel("f(x)")
    
    # Set reasonable y-limits for PDFs
    if i == 0:  # sqrt function has singularity at 0
        axs[i, 1].set_ylim(0, 5)
    else:
        axs[i, 1].set_ylim(0, max(f_pdf(x)) * 1.2)

plt.tight_layout()
plt.show()
```

All of these graphs satisfy being continous, non-decreasing, and is bounded between 0 and 1. Thus, these are CDFs.

## Density Functions (Derivatives of CDFs)

### 1. $F(x) = \begin{cases} 0, & x \le 0 \\ \sqrt{x}, & 0 \le x \le 1 \\ 1, & x \ge 1 \end{cases}$

**Derivative:**
$$f(x) = F'(x) = \begin{cases}
0, & x \le 0 \\
\frac{1}{2\sqrt{x}}, & 0 < x < 1 \\
0, & x \ge 1
\end{cases}$$

**Calculation:** For $0 < x < 1$: $\frac{d}{dx}\sqrt{x} = \frac{d}{dx}x^{1/2} = \frac{1}{2}x^{-1/2} = \frac{1}{2\sqrt{x}}$

---

### 2. $F(x) = \dfrac{1}{1+e^{-x}}$ 

**Derivative:**
$$f(x) = F'(x) = \frac{e^{x}}{(1+e^{x})^2}$$

**Calculation:** First rewrite the CDF with positive exponents by multiplying numerator and denominator by $e^x$:
$$F(x) = \frac{1}{1+e^{-x}} \cdot \frac{e^x}{e^x} = \frac{e^x}{e^x + 1}$$

Now apply the quotient rule:
$$f(x) = \frac{d}{dx}\left[\frac{e^x}{1+e^x}\right] = \frac{e^x(1+e^x) - e^x \cdot e^x}{(1+e^x)^2} = \frac{e^x + e^{2x} - e^{2x}}{(1+e^x)^2} = \frac{e^x}{(1+e^x)^2}$$

---

### 3. For $a < c < b$:
$$F(x) = \begin{cases}
0, & x \le a \\
\frac{(x-a)^2}{(b-a)(c-a)}, & a \le x \le c \\
1 - \frac{(b-x)^2}{(b-a)(b-c)}, & c < x < b \\
1, & x \ge b
\end{cases}$$

**Derivative:**
$$f(x) = F'(x) = \begin{cases}
0, & x \le a \\
\frac{2(x-a)}{(b-a)(c-a)}, & a < x < c \\
\frac{2(b-x)}{(b-a)(b-c)}, & c < x < b \\
0, & x \ge b
\end{cases}$$

**Calculations:**
- For $a < x < c$: $\frac{d}{dx}\left[\frac{(x-a)^2}{(b-a)(c-a)}\right] = \frac{2(x-a)}{(b-a)(c-a)}$
- For $c < x < b$: $\frac{d}{dx}\left[1 - \frac{(b-x)^2}{(b-a)(b-c)}\right] = \frac{2(b-x)}{(b-a)(b-c)}$

---

### 4. $F(x) = \begin{cases} 0, & x \le 0 \\ 1 - e^{-x}, & x > 0 \end{cases}$ (Exponential Distribution)

**Derivative:**
$$f(x) = F'(x) = \begin{cases}
0, & x \le 0 \\
e^{-x}, & x > 0
\end{cases}$$

**Calculation:** For $x > 0$: $\frac{d}{dx}\left[1 - e^{-x}\right] = 0 - (-e^{-x}) = e^{-x}$

---

### 5. $F(x) = \begin{cases} 0, & x \le 0 \\ x, & 0 < x < 1\\ 1, & x \ge 1 \end{cases}$ (Uniform Distribution)

**Derivative:**
$$f(x) = F'(x) = \begin{cases}
0, & x \le 0 \\
1, & 0 < x < 1 \\
0, & x \ge 1
\end{cases}$$

**Calculation:** For $0 < x < 1$: $\frac{d}{dx}[x] = 1$

$$\frac{d}{dx}\left[\frac{1}{1+e^{-x}}\right] = \frac{d}{dx}\left[(1+e^{-x})^{-1}\right] = -(1+e^{-x})^{-2} \cdot (-e^{-x}) = \frac{e^{-x}}{(1+e^{-x})^2}$$

**Alternative form:** $f(x) = \frac{e^x}{(1+e^x)^2}$

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import scipy as sp

# Parameters
mu = 0
sigmas = np.arange(1, 5.5, 0.5)  # [1, 1.5, 2, 2.5, 3, 3.5, 4, 4.5, 5]
grid = np.linspace(-20, 20, 400)

# Plot setup
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for s in sigmas:
    pdf = sp.stats.logistic.pdf(grid, loc=mu, scale=s)
    cdf = sp.stats.logistic.cdf(grid, loc=mu, scale=s)
    
    sns.lineplot(x=grid, y=pdf, ax=axes[0], label=f'σ={s:.1f}')
    sns.lineplot(x=grid, y=cdf, ax=axes[1], label=f'σ={s:.1f}')

axes[0].set_title('Logistic PDFs for Different σ')
axes[0].set_xlabel('x')
axes[0].set_ylabel('Density')

axes[1].set_title('Logistic CDFs for Different σ')
axes[1].set_xlabel('x')
axes[1].set_ylabel('Probability')

plt.tight_layout()
plt.show()
```


### Negative Binomial

**Support**

Looking at the CDF and PMF of this distribution, it's clear that the support is K = {0, 1, 2, 3, ...}, where K is the number of failures before a success and K represents all non-negative integers (integers ≥ 0).




**Density Distribution Plot**


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy as sp

# Negative Binomial parameters
# Parameter sets (3 for each)
r_values = [3, 4, 5, 8]        # number of successes
p_values = [0.2, 0.3, 0.4, 0.6]  # probability of success

# Support (integer x-values)
grid = np.arange(0, 40, 1)

# Create subplots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Loop over parameter combinations
for r, p in zip(r_values, p_values):
    # Compute PMF and CDF
    pmf = sp.stats.nbinom.pmf(grid, n=r, p=p)
    cdf = sp.stats.nbinom.cdf(grid, n=r, p=p)

    # Plot PMF
    sns.lineplot(x=grid, y=pmf, ax=axes[0], label=f"r={r}, p={p}")
    # Plot CDF
    sns.lineplot(x=grid, y=cdf, ax=axes[1], label=f"r={r}, p={p}")

# Titles and labels
axes[0].set_title("Negative Binomial PMF (Density)")
axes[0].set_xlabel("k (Number of Failures)")
axes[0].set_ylabel("Probability")

axes[1].set_title("Negative Binomial CDF (Distribution)")
axes[1].set_xlabel("k (Number of Failures)")
axes[1].set_ylabel("Cumulative Probability")

plt.tight_layout()
plt.show()
```


### Categorical
The support is **K = {1, 2, 3, ..., k}**, where k is the total number of categories.
- Each value represents one of the k possible categories
- For example, 5 categories, K ∈ {1, 2, 3, 4, 5}

**Density Distribution Plot**


```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

categories = ["W", "X", "Y", "Z"]
distributions = { "Equal_Split": [0.25, 0.25, 0.25, 0.25], "High_W": [0.5, 0.2, 0.2, 0.1], "High_Z": [0.1, 0.2, 0.2, 0.5], "Middle_Weighted": [0.1, 0.4, 0.4, 0.1]
}

pmf_data = []
for name, probs in distributions.items():
    for cat, p in zip(categories, probs):
        pmf_data.append({"Distribution": name, "Category": cat, "Probability": p})
pmf_df = pd.DataFrame(pmf_data)

cdf_df = pmf_df.copy()
cdf_df["Cumulative"] = cdf_df.groupby("Distribution")["Probability"].cumsum()

plt.figure(figsize=(8,4))
sns.barplot(data=pmf_df, x="Category", y="Probability", hue="Distribution")
plt.title("Categorical PMF")
plt.ylim(0,1)
plt.ylabel("Probability")
plt.show()

plt.figure(figsize=(8,4))
sns.barplot(data=cdf_df, x="Category", y="Cumulative", hue="Distribution")
plt.title("Categorical CDF")
plt.ylim(0,1.05)
plt.ylabel("Cumulative Probability")
plt.show()
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Categories and distribution
categories = ["W", "X", "Y", "Z"]
probs = [0.5, 0.2, 0.2, 0.1]  # High_W distribution

np.random.seed(42)
sample = np.random.choice(categories, size=1000, p=probs)

def ecdf(data):
    xs = np.sort(data)
    ys = np.arange(1, len(xs)+1) / len(xs)
    return xs, ys

xs_sample, ys_sample = ecdf(sample)

cdf_theory = np.cumsum(probs)

plt.figure(figsize=(8,4))
# Sample ECDF
plt.step(xs_sample, ys_sample, where='post', label='Sample ECDF')
# Theoretical CDF
plt.step(categories, cdf_theory, where='post', color='red', marker='o', label='Theoretical CDF')
plt.xlabel("Category")
plt.ylabel("Cumulative Probability")
plt.title("ECDF vs Theoretical CDF (High_W)")
plt.legend()
plt.show()

sample_numeric = [categories.index(s) for s in sample]

plt.figure(figsize=(8,4))
# Sample KDE
sns.kdeplot(sample_numeric, bw_adjust=0.5, fill=True, label='Sample KDE')
# Theoretical PMF as markers
plt.scatter(range(len(categories)), probs, color='red', s=100, zorder=5, label='Theoretical PMF')
plt.xticks(ticks=range(len(categories)), labels=categories)
plt.xlabel("Category")
plt.ylabel("Density / Probability")
plt.title("KDE vs Theoretical PMF (High_W)")
plt.legend()
plt.show()
```
